```python
total_bookings = len(df)

cancelled_bookings = df["is_canceled"].sum()

cancellation_rate = (cancelled_bookings / total_bookings) * 100

print("Total bookings:", total_bookings)
print("Cancelled bookings:", cancelled_bookings)
print("Cancellation rate:", round(cancellation_rate, 2), "%")
```

```python
hotel_cancellation = df.groupby("hotel")["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

hotel_cancellation["cancellation_rate"] = (
    hotel_cancellation["cancelled_bookings"]
    / hotel_cancellation["total_bookings"]
) * 100

hotel_cancellation["cancellation_rate"] = hotel_cancellation[
    "cancellation_rate"
].round(2)

print(hotel_cancellation)
```

```python
channel_cancellation = df.groupby("distribution_channel")["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

channel_cancellation["cancellation_rate"] = (
    channel_cancellation["cancelled_bookings"]
    / channel_cancellation["total_bookings"]
) * 100

channel_cancellation["cancellation_rate"] = (
    channel_cancellation["cancellation_rate"].round(2)
)

print(channel_cancellation.sort_values(
    "cancellation_rate",
    ascending=False
))
```

```python
segment_customer_analysis = df.groupby(
    ["market_segment", "customer_type"]
)["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

segment_customer_analysis["cancellation_rate"] = (
    segment_customer_analysis["cancelled_bookings"]
    / segment_customer_analysis["total_bookings"]
) * 100

segment_customer_analysis["cancellation_rate"] = (
    segment_customer_analysis["cancellation_rate"].round(2)
)

print("Cancellation Analysis by Market Segment and Customer Type:")
print(
    segment_customer_analysis.sort_values(
        "cancellation_rate",
        ascending=False
    )
)
```

```python
room_deposit_analysis = df.groupby(
    ["reserved_room_type", "deposit_type"]
)["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

room_deposit_analysis["cancellation_rate"] = (
    room_deposit_analysis["cancelled_bookings"]
    / room_deposit_analysis["total_bookings"]
) * 100

room_deposit_analysis["cancellation_rate"] = (
    room_deposit_analysis["cancellation_rate"].round(2)
)

print("Cancellation Analysis by Room Type and Deposit Type:")
print(
    room_deposit_analysis.sort_values(
        "cancellation_rate",
        ascending=False
    )
)
```

```python
# STEP 53 - MASTER CANCELLATION ANALYSIS

analysis_columns = [
    "hotel",
    "market_segment",
    "distribution_channel",
    "deposit_type",
    "customer_type",
    "reserved_room_type",
    "meal",
    "lead_time_group",
    "adr_group",
    "is_repeated_guest"
]

results = []

for column in analysis_columns:

    analysis = df.groupby(column, observed=False)["is_canceled"].agg(
        total_bookings="count",
        cancelled_bookings="sum"
    )

    analysis["cancellation_rate"] = (
        analysis["cancelled_bookings"]
        / analysis["total_bookings"]
    ) * 100

    for category, row in analysis.iterrows():
        results.append({
            "Factor": column,
            "Category": category,
            "Total Bookings": row["total_bookings"],
            "Cancelled Bookings": row["cancelled_bookings"],
            "Cancellation Rate": round(row["cancellation_rate"], 2)
        })

master_analysis = pd.DataFrame(results)

master_analysis = master_analysis.sort_values(
    "Cancellation Rate",
    ascending=False
)

print("MASTER CANCELLATION ANALYSIS")
print("=" * 60)

print("\nHighest cancellation patterns:")
print(master_analysis.head(20).to_string(index=False))

```

```python
# STEP 54 - FIND STRONGEST CANCELLATION FACTORS

factor_summary = master_analysis.groupby("Factor").agg(
    average_cancellation_rate=("Cancellation Rate", "mean"),
    highest_cancellation_rate=("Cancellation Rate", "max"),
    total_bookings=("Total Bookings", "sum")
)

factor_summary = factor_summary.sort_values(
    "average_cancellation_rate",
    ascending=False
)

factor_summary["average_cancellation_rate"] = (
    factor_summary["average_cancellation_rate"].round(2)
)

factor_summary["highest_cancellation_rate"] = (
    factor_summary["highest_cancellation_rate"].round(2)
)

print("STRONGEST CANCELLATION FACTORS")
print("=" * 60)
print(factor_summary)
```

```python
# STEP 55 - CANCELLATION RISK SCORE

df["risk_score"] = 0

# Higher lead time can indicate higher cancellation risk
df.loc[df["lead_time"] > 90, "risk_score"] += 2

# Previous cancellations increase risk
df.loc[df["previous_cancellations"] > 0, "risk_score"] += 2

# No deposit can increase risk
df.loc[df["deposit_type"] == "No Deposit", "risk_score"] += 1

# Booking changes can indicate booking uncertainty
df.loc[df["booking_changes"] > 0, "risk_score"] += 1

# No previous successful bookings
df.loc[df["previous_bookings_not_canceled"] == 0, "risk_score"] += 1

# Convert score into risk level
df["risk_level"] = "Low Risk"

df.loc[
    df["risk_score"].between(2, 3),
    "risk_level"
] = "Medium Risk"

df.loc[
    df["risk_score"] >= 4,
    "risk_level"
] = "High Risk"

print("Cancellation Risk Summary:")
print(df["risk_level"].value_counts())

print("\nRisk Score Examples:")
print(
    df[
        [
            "risk_score",
            "risk_level",
            "lead_time",
            "previous_cancellations",
            "deposit_type",
            "booking_changes",
            "previous_bookings_not_canceled"
        ]
    ].head(10)
)
```

```python
# STEP 56 - VALIDATE RISK LEVELS

risk_validation = df.groupby("risk_level")["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

risk_validation["actual_cancellation_rate"] = (
    risk_validation["cancelled_bookings"]
    / risk_validation["total_bookings"]
) * 100

risk_validation["actual_cancellation_rate"] = (
    risk_validation["actual_cancellation_rate"].round(2)
)

# Put risk levels in logical order
risk_order = ["Low Risk", "Medium Risk", "High Risk"]

risk_validation = risk_validation.reindex(risk_order)

print("RISK LEVEL VALIDATION")
print("=" * 60)
print(risk_validation)
```

```python
# STEP 57 - HIGH RISK BOOKING ANALYSIS

high_risk = df[df["risk_level"] == "High Risk"]

print("HIGH RISK BOOKING SUMMARY")
print("=" * 60)

print("Total High Risk Bookings:", len(high_risk))

print("\nHigh Risk by Deposit Type:")
print(high_risk["deposit_type"].value_counts())

print("\nHigh Risk by Customer Type:")
print(high_risk["customer_type"].value_counts())

print("\nHigh Risk by Market Segment:")
print(high_risk["market_segment"].value_counts())

print("\nHigh Risk by Hotel:")
print(high_risk["hotel"].value_counts())
```

```python
# STEP 58 - HIGH RISK CANCELLATION RATE

high_risk_cancelled = high_risk["is_canceled"].sum()
high_risk_total = len(high_risk)

high_risk_cancellation_rate = (
    high_risk_cancelled / high_risk_total
) * 100

print("HIGH RISK CANCELLATION ANALYSIS")
print("=" * 60)

print("Total High Risk Bookings:", high_risk_total)
print("High Risk Bookings Cancelled:", high_risk_cancelled)
print(
    "High Risk Cancellation Rate:",
    round(high_risk_cancellation_rate, 2),
    "%"
)
```

```python
# STEP 59 - OVERALL VS HIGH RISK CANCELLATION

overall_cancellation_rate = (
    df["is_canceled"].mean()
) * 100

high_risk_rate = (
    high_risk["is_canceled"].mean()
) * 100

risk_lift = high_risk_rate / overall_cancellation_rate

print("CANCELLATION RATE COMPARISON")
print("=" * 60)

print(
    "Overall Cancellation Rate:",
    round(overall_cancellation_rate, 2),
    "%"
)

print(
    "High Risk Cancellation Rate:",
    round(high_risk_rate, 2),
    "%"
)

print(
    "High Risk Cancellation Lift:",
    round(risk_lift, 2),
    "x"
)
```

```python
# STEP 60 - HIGH RISK DRIVER ANALYSIS

print("HIGH RISK DRIVER ANALYSIS")
print("=" * 60)

print("\n1. Lead Time:")
print(high_risk["lead_time"].describe())

print("\n2. Previous Cancellations:")
print(high_risk["previous_cancellations"].value_counts())

print("\n3. Deposit Type:")
print(high_risk["deposit_type"].value_counts())

print("\n4. Booking Changes:")
print(high_risk["booking_changes"].value_counts())

print("\n5. Previous Successful Bookings:")
print(high_risk["previous_bookings_not_canceled"].value_counts())
```

```python
# STEP 61 - CANCELLATION RATE BY RISK LEVEL

risk_analysis = df.groupby("risk_level")["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

risk_analysis["cancellation_rate"] = (
    risk_analysis["cancelled_bookings"]
    / risk_analysis["total_bookings"]
) * 100

risk_analysis["cancellation_rate"] = (
    risk_analysis["cancellation_rate"].round(2)
)

risk_analysis = risk_analysis.reindex(
    ["Low Risk", "Medium Risk", "High Risk"]
)

print("CANCELLATION RATE BY RISK LEVEL")
print("=" * 60)
print(risk_analysis)
```

```python
# STEP 62 - BUSINESS DECISION CATEGORY

df["decision"] = "Monitor"

df.loc[
    df["risk_level"] == "Medium Risk",
    "decision"
] = "Follow Up"

df.loc[
    df["risk_level"] == "High Risk",
    "decision"
] = "Priority Follow Up"

print("BUSINESS DECISION SUMMARY")
print("=" * 60)

print(df["decision"].value_counts())
```

```python
# STEP 63 - LIKELY CANCELLATION DRIVER

df["likely_driver"] = "Normal Booking Pattern"

df.loc[
    df["previous_cancellations"] > 0,
    "likely_driver"
] = "Previous Cancellation History"

df.loc[
    (df["previous_cancellations"] == 0) &
    (df["lead_time"] > 90),
    "likely_driver"
] = "Long Lead Time"

df.loc[
    (df["previous_cancellations"] == 0) &
    (df["lead_time"] <= 90) &
    (df["deposit_type"] == "No Deposit"),
    "likely_driver"
] = "No Deposit"

df.loc[
    (df["previous_cancellations"] == 0) &
    (df["lead_time"] <= 90) &
    (df["deposit_type"] != "No Deposit") &
    (df["booking_changes"] > 0),
    "likely_driver"
] = "Booking Changes"

print("LIKELY CANCELLATION DRIVER")
print("=" * 60)

print(df["likely_driver"].value_counts())
```

```python
# STEP 64 - RECOMMENDED BUSINESS ACTION

df["recommended_action"] = "Normal Monitoring"

df.loc[
    df["likely_driver"] == "Previous Cancellation History",
    "recommended_action"
] = "Send confirmation and cancellation-policy reminder"

df.loc[
    df["likely_driver"] == "Long Lead Time",
    "recommended_action"
] = "Send confirmation reminder closer to arrival"

df.loc[
    df["likely_driver"] == "No Deposit",
    "recommended_action"
] = "Send confirmation and consider deposit/payment verification"

df.loc[
    df["likely_driver"] == "Booking Changes",
    "recommended_action"
] = "Contact guest to confirm booking details"

# Display sample decisions
print("RECOMMENDED BUSINESS ACTIONS")
print("=" * 60)

print(
    df[
        [
            "risk_level",
            "likely_driver",
            "decision",
            "recommended_action"
        ]
    ].head(15).to_string(index=False)
)
```

```python
# STEP 65 - ESTIMATED VALUE AT RISK

df["total_nights"] = (
    df["stays_in_weekend_nights"]
    + df["stays_in_week_nights"]
)

df["estimated_booking_value"] = (
    df["adr"] * df["total_nights"]
)

high_risk_value = df.loc[
    df["risk_level"] == "High Risk",
    "estimated_booking_value"
].sum()

cancelled_high_risk_value = df.loc[
    (df["risk_level"] == "High Risk") &
    (df["is_canceled"] == 1),
    "estimated_booking_value"
].sum()

print("ESTIMATED VALUE AT RISK")
print("=" * 60)

print(
    "Estimated value of High Risk bookings: ₹",
    round(high_risk_value, 2)
)

print(
    "Estimated value of cancelled High Risk bookings: ₹",
    round(cancelled_high_risk_value, 2)
)
```

```python
# STEP 66 - FINAL DECISION SUPPORT SUMMARY

final_summary = df[
    [
        "hotel",
        "lead_time",
        "deposit_type",
        "previous_cancellations",
        "booking_changes",
        "risk_score",
        "risk_level",
        "likely_driver",
        "decision",
        "recommended_action",
        "estimated_booking_value"
    ]
].copy()

print("FINAL DECISION SUPPORT SUMMARY")
print("=" * 80)

print(final_summary.head(20).to_string(index=False))
```

```python
# STEP 73 - LIKELY DRIVER SUMMARY

driver_summary = df.groupby("likely_driver")["is_canceled"].agg(
    total_bookings="count",
    cancelled_bookings="sum"
)

driver_summary["cancellation_rate"] = (
    driver_summary["cancelled_bookings"]
    / driver_summary["total_bookings"]
) * 100

driver_summary["cancellation_rate"] = (
    driver_summary["cancellation_rate"].round(2)
)

driver_summary = driver_summary.sort_values(
    "cancellation_rate",
    ascending=False
)

print("LIKELY CANCELLATION DRIVER SUMMARY")
print("=" * 60)
print(driver_summary)
```

```python
# STEP 74 - BUSINESS ACTION SUMMARY

action_summary = df.groupby(
    ["likely_driver", "recommended_action"]
).agg(
    total_bookings=("is_canceled", "count"),
    cancelled_bookings=("is_canceled", "sum")
).reset_index()

action_summary["cancellation_rate"] = (
    action_summary["cancelled_bookings"]
    / action_summary["total_bookings"]
) * 100

action_summary["cancellation_rate"] = (
    action_summary["cancellation_rate"].round(2)
)

action_summary = action_summary.sort_values(
    "cancellation_rate",
    ascending=False
)

print("BUSINESS ACTION SUMMARY")
print("=" * 80)

print(action_summary.to_string(index=False))
```

```python
# STEP 75 - FINAL BUSINESS IMPACT SUMMARY

high_risk_bookings = df[df["risk_level"] == "High Risk"]

total_high_risk = len(high_risk_bookings)

high_risk_cancelled = high_risk_bookings["is_canceled"].sum()

high_risk_cancellation_rate = (
    high_risk_cancelled / total_high_risk
) * 100

estimated_value_at_risk = (
    high_risk_bookings["estimated_booking_value"].sum()
)

estimated_cancelled_value = (
    high_risk_bookings.loc[
        high_risk_bookings["is_canceled"] == 1,
        "estimated_booking_value"
    ].sum()
)

print("FINAL BUSINESS IMPACT SUMMARY")
print("=" * 70)

print("High Risk Bookings:", total_high_risk)

print(
    "High Risk Bookings Cancelled:",
    high_risk_cancelled
)

print(
    "High Risk Cancellation Rate:",
    round(high_risk_cancellation_rate, 2),
    "%"
)

print(
    "Estimated Booking Value at Risk:",
    round(estimated_value_at_risk, 2)
)

print(
    "Estimated Value of Cancelled High Risk Bookings:",
    round(estimated_cancelled_value, 2)
)
```
